```python
# 11. Ask the user for a number and check whether it is prime.
n=int(input('Enter the number : '))
a=[]
if n<=1:
  print(n,'is not prime')
else:
  for i in range(2,int(n ** 0.5) + 1):
    if n%i==0:
      a.append(i)
      break
  #print(a)
  if len(a)==0:
    print(n,'is prime')
  else:
    print(n,'is not prime')
```

```python
# 14. Build a number-guessing game. The program picks a secret number from 1 to 10
#     (use import random and random.randint(1, 10)). The user keeps guessing until correct,
#     and the program says “too high” or “too low” after each wrong guess. At the end, print how many attempts it took.
import random
a=random.randint(1,10)
n=int(input('guess the number : '))
count=0
while n!=a:
  if n>a:
    print('too high')
    count=count+1
    n=int(input('guess the number : '))

  elif n<a:
    print('too low')
    count=count+1
    n=int(input('guess the number : '))
  else:
    print('congratulations!!! your guess is correct - ',a)
    count=count+1

else:
    print('congratulations!!! your guess is correct - ',a)
    count=count+1
print('You got it in ',count,' attempts')
```
